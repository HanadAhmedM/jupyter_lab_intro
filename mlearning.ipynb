```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('penguins')

df.head()
df.info()

plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x='species',
    hue='species',
    palette='viridis',
    legend=False
)
plt.title('Antal pingviner per art')
plt.xlabel('Pingvinart')
plt.ylabel('Antal')
plt.show()
```

```python
df_clean= df.dropna()
numric_df = df_clean.select_dtypes(
    include=['float64','int64'])
corr_matrix = numric_df.corr()
plt.figure(figsize=(8, 5))
sns.heatmap(
    corr_matrix,
    annot= True,
    vmin=1,
    vmax=1,
    cmap='coolwarm'
    );
```
