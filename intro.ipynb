```python
import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

column_names = [
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width',
    'species'
]

iris_df = pd.read_csv(url, names=column_names)

print(iris_df.head())
```

```python
iris_df.describe(include='all')
```

```python
mean_speal_lenght = iris_df['sepal_length'].mean()
print(mean_speal_lenght)
```

```python
mean_per_species= iris_df.groupby('species').mean()
mean_per_species
```

```python
filtered_df = iris_df[iris_df['petal_length'] > 3.0]
print(filtered_df.head())
```
