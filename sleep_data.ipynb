```python

import kagglehub
from kagglehub import KaggleDatasetAdapter

file_path = "dataset_2191_sleep.csv"

df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "mathurinache/sleep-dataset",
    file_path
)

print("First 5 records:")
print(df.head())


```

```python

import pandas as pd

df["total_sleep"] = pd.to_numeric(
    df["total_sleep"],
    errors="coerce"
)

filter_df = df[df["total_sleep"] > 5.3]

print(filter_df.head())
```
