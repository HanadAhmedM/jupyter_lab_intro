```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("banana_quality.csv")
df.head()
```

```python
df.info()
```

```python
df.isna().sum().sum()
```

```python
df.duplicated().sum()
```

```python
df.describe().T.round(2)
```

```python
df["Quality"].value_counts()
```

```python
sns.countplot(data=df,x="Quality", hue="Quality");
```

```python
long_df= df.melt(id_vars="Quality", var_name="feature", value_name="value")
long_df.head()
```

```python
sns.displot(data= long_df, x="value", col="feature", col_wrap=4 , hue="Quality", kde=True);
```

```python
sns.catplot(data=long_df, x= "Quality", y="value",col ="feature", hue="Quality", col_wrap=4, )
```

```python
corr=df.assign(
    Quality=(df["Quality"] == "Good").astype(int)
)
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr,
    annot=True,
    vmin=-1,
    vmax=1,
    cmap="coolwarm"
)
```

```python
sns.pairplot(df.sample(100,random_state=42),hue="Quality", corner=True, height= 1.0);
```
